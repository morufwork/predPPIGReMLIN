```python
import os
import math

def split_graph_file(input_file, output_dir, num_splits=10):
    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)

    # Read the entire file
    with open(input_file, 'r') as f:
        content = f.read()

    # Split graphs using '#' delimiter
    graphs = [g.strip() for g in content.split('#') if g.strip()]

    total_graphs = len(graphs)
    print(f"Total graphs found: {total_graphs}")

    # Compute size of each split
    chunk_size = math.ceil(total_graphs / num_splits)

    # Split and save
    for i in range(num_splits):
        start = i * chunk_size
        end = start + chunk_size
        chunk = graphs[start:end]

        if not chunk:
            break

        output_file = os.path.join(output_dir, f"train_part_{i+1}.txt")

        with open(output_file, 'w') as f:
            f.write("#".join(chunk) + "#")

        print(f"Saved {len(chunk)} graphs to {output_file}")

    print("Splitting completed!")


# Usage
input_file = "masif_training_graphs.txt"
output_dir = "masif_train"

split_graph_file(input_file, output_dir, num_splits=10)
```
